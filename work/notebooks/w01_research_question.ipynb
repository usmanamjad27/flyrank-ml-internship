My provisional lane is Refresh / Content Opportunity Scoring. I want to investigate which content pages should be reviewed first using their search performance and content-related measurements. This could help content editors prioritize their limited time. I will start with simple rules and compare other approaches only if the data supports doing so.

## 2. The question: decision, action, cost of a wrong call

Research question: Which content pages should an SEO or content editor review first, based on their available search performance and content measurements?

Unit of analysis: One content page.

Decision: Which pages should receive review priority?

Output: A ranked list of pages that may deserve attention.

Action: An editor reviews a prioritized page and decides whether it needs a content update, further investigation, or no change.

Cost of a wrong recommendation: A false positive wastes editorial time on a page that does not need attention. A false negative may cause the team to miss a page worth reviewing.

Why data or ML can help: Combining several measurements may help prioritize pages more consistently than relying on intuition alone. I will compare the results against a simple baseline and evaluate whether the ranking is useful.

## 3. Quick look at the data (2-3 real numbers)

I will examine the starter dataset's size, the proportion of pages currently labeled as declining, and the distribution of search impressions. These measurements will help me determine whether the dataset provides enough information to investigate content-review prioritization. The current decline label is an observed signal, not proof that a page will continue declining or that updating it will improve performance.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

# Load the starter dataset from the public repository.
url = (
    "https://raw.githubusercontent.com/"
    "usmanamjad27/flyrank-ml-internship/main/"
    "data/raw/content_refresh_anonymized.csv"
)

df = pd.read_csv(url)

# Check that the required columns exist.
required_columns = ["trend_direction", "impressions_90d"]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# Calculate real numbers from the dataset.
total_pages = len(df)

declining_rate = (
    df["trend_direction"].eq("down").mean() * 100
)

impressions = pd.to_numeric(
    df["impressions_90d"], errors="coerce"
)

median_impressions = impressions.median()

print(f"Total pages in dataset: {total_pages:,}")
print(f"Pages labeled as declining: {declining_rate:.2f}%")
print(f"Median impressions over 90 days: {median_impressions:,.0f}")

```
